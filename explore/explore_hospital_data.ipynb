```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load dataset
data = pd.read_csv("./data/hospital.csv")
```

# Inspect Data

```python
data.head()
```

```python
# inspect data type and shape

print("df type:", type(data))
print("df shape:", data.shape)
```

```python
# check unique values
data.nunique()
```

```python
# check missing values

data.isnull().sum()
```

```python
# visualize missing values

plt.figure(figsize=(12,6))
sns.heatmap(data.isnull(), cbar=True, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.show()
```

```python
data.duplicated().sum()
```

# Exploratory Data Analysis

```python
data["visit_date"] = pd.to_datetime(data["visit_date"], errors="coerce")
data["admission_date"] = pd.to_datetime(data["admission_date"], errors="coerce")
```

#### Patient demographics

```python
gender_counts = data["gender"].value_counts()

gender_counts
```

```python
gender_counts.plot(kind="bar")
plt.title("Patients by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.show()
```

```python
subset = data[[ "age", "sickness", "treatment_status"]]
subset.head()
```

```python
subset['age'].hist(bins=20, edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.title('Age Distribution')
plt.show()
```

```python
subset['treatment_status'].value_counts().plot(kind='bar', edgecolor='black')
plt.ylabel('Count')
plt.title('Treatment Status Counts')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# 1. Age distribution
sns.histplot(subset['age'], bins=20, kde=True, ax=axes[0])
axes[0].set_title('Age Distribution')

# 2. Age by treatment status
sns.boxplot(data=subset, x='treatment_status', y='age', ax=axes[1])
axes[1].set_title('Age by Treatment Status')
axes[1].tick_params(axis='x', rotation=30)

# 3. Treatment status counts
subset['treatment_status'].value_counts().plot(kind='bar', ax=axes[2], edgecolor='black')
axes[2].set_title('Treatment Status Counts')
axes[2].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

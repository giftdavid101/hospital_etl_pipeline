# EDA

```python
hospital_data["visit_date"] = pd.to_datetime(hospital_data["visit_date"], errors="coerce")
hospital_data["admission_date"] = pd.to_datetime(hospital_data["admission_date"], errors="coerce")
```

#### Patient demographics

```python
gender_counts = hospital_data["gender"].value_counts()

gender_counts
```

```python
gender_counts.plot(kind="bar")
plt.title("Patients by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.show()
```

```python
hospital_data.notna()
```

```python
df_complete = hospital_data[
    [
        "patient_id",
        "age",
        "sickness",
        "doctor_report",
        "treatment_status"
    ]
]
```

```python
df_complete = hospital_data.loc[:, hospital_data.notna().all()]
df_complete 
```

```python
subset = df_complete[[ "age", "sickness", "treatment_status"]]
subset.head()
```

```python


subset['age'].hist(bins=20, edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.title('Age Distribution')
plt.show()
```

```python
subset['treatment_status'].value_counts().plot(kind='bar', edgecolor='black')
plt.ylabel('Count')
plt.title('Treatment Status Counts')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# 1. Age distribution
sns.histplot(subset['age'], bins=20, kde=True, ax=axes[0])
axes[0].set_title('Age Distribution')

# 2. Age by treatment status
sns.boxplot(data=subset, x='treatment_status', y='age', ax=axes[1])
axes[1].set_title('Age by Treatment Status')
axes[1].tick_params(axis='x', rotation=30)

# 3. Treatment status counts
subset['treatment_status'].value_counts().plot(kind='bar', ax=axes[2], edgecolor='black')
axes[2].set_title('Treatment Status Counts')
axes[2].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```
